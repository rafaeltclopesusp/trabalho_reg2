```python
file_bag = pysus.ftp.sih(state="RN", year=2024, month=[1])
```

Aquí están los archivos que fueron descargados y se encuentran en `file_bag`:

```python
# Para ver los archivos individuales dentro de file_bag
print("Archivos disponibles en file_bag:")
for i, file_obj in enumerate(file_bag):
    print(f"{i}: {file_obj.name}")
```

Para cargar solo uno de estos archivos, puedes seleccionar el archivo deseado del `file_bag` por su índice (por ejemplo, `file_bag[0]` para el primero, `file_bag[1]` para el segundo, etc.) y luego pasarlo a `pysus.read_sih`.

```python
# Cargar solo el primer archivo en un nuevo DataFrame
# Puedes cambiar el índice (0, 1, 2, 3) para seleccionar un archivo diferente
selected_file_object = file_bag[3]
file_path = selected_file_object.path
df_single_file = pd.read_parquet(file_path)

print(f"Cargado el archivo: {selected_file_object.name}")
# Mostrar las primeras filas del DataFrame del archivo único
display(df_single_file.head())
```

```python
import seaborn as sns
```

```python
df2 = df1[['IDADE','DIAS_PERM','US_TOT','MORTE']]
matriz_corr = df2.corr()

# 2. Configure o tamanho da figura (opcional)
plt.figure(figsize=(10, 8))

# 3. Crie o heatmap do Seaborn
sns.heatmap(
    matriz_corr,
    annot=True,  # Mostra os números dentro dos quadrados
    fmt=".2f",  # Limita os números a 2 casas decimais
    cmap="coolwarm",  # Paleta de cores (azul para negativo, vermelho para positivo)
    vmin=-1,  # Valor mínimo da escala (correlação negativa perfeita)
    vmax=1,  # Valor máximo da escala (correlação positiva perfeita)
    linewidths=0.5,  # Linha fina separando os quadrados
)

# 4. Exiba o gráfico
plt.show()
```
